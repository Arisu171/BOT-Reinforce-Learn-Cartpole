This notebook guides you through:
1. Exploring the observation (state) space and action space.
2. Executing a sample episode with the Random policy.
3. Evaluating the Random Baseline over 500 episodes and visualizing the return distribution.

```python
import os
import sys
sys.path.insert(0, os.path.abspath('..'))
os.chdir('..')

import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
from libraries.envs import make_env
from libraries.policies import get_policy
from libraries.evaluate import evaluate_policy
```

## 1. State and Action Spaces

CartPole-v1 observation vector consists of 4 continuous variables:
- $x$: Cart position
- $\dot{x}$: Cart velocity
- $\theta$: Pole angle
- $\dot{\theta}$: Pole angular velocity

Actions are discrete: `0` (push cart left) or `1` (push cart right).

```python
env = make_env('v0', seed=42)
obs, info = env.reset(seed=42)
print('Observation space:', env.observation_space)
print('Action space:', env.action_space)
print('Initial observation:', obs)
env.close()
```

## 2. Test Single Episode with Random Policy

```python
env = make_env('v0', seed=42)
obs, info = env.reset(seed=42)
total_reward = 0.0
steps = 0
done = False

while not done:
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    steps += 1
    done = terminated or truncated

print(f'Episode finished after {steps} steps. Total return = {total_reward}')
env.close()
```

## 3. Evaluate Random Baseline (500 episodes)

Theoretically, a random policy on CartPole-v1 achieves an average return around 20-25 points.

```python
policy_random = get_policy('random', seed=123)
eval_res = evaluate_policy(policy_random, n_episodes=500, seed_start=100)
print(f'Mean return: {eval_res["mean"]:.2f}')
print(f'Std:         {eval_res["std"]:.2f}')
print(f'Min / Max:   {eval_res["min"]:.1f} / {eval_res["max"]:.1f}')
```

## 4. Episode Length Distribution of Random Baseline

```python
returns = eval_res['returns']
plt.figure(figsize=(8, 4.5))
plt.hist(returns, bins=25, color='steelblue', edgecolor='black', alpha=0.8)
plt.axvline(eval_res['mean'], color='red', linestyle='--', linewidth=2, label=f'Mean = {eval_res["mean"]:.1f}')
plt.title('Return Distribution of Random Policy (500 episodes)')
plt.xlabel('Return')
plt.ylabel('Episode count')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
